```python
#1. Create a NumPy array containing the numbers from 1 to 20.
#a. Reshape it into a 4 × 5 array.
#b. Display the last two rows of the reshaped array
import numpy as np
s=np.arange(1,21)
a=s.reshape(4,5)
b=a[-2::]
print (a)
print (b)
```

```python
#2. Given the following array,
#arr = np.array([12, 45, 7, 23, 56, 18, 34, 9, 62, 27])
#a. Display all elements that are greater than 20 and less than 60.
#b. Display the positions of these elements in the original array
r = np.array([12, 45, 7, 23, 56, 18, 34, 9, 62, 27])
a=r[(r>20)&(r<60)]
b=np.where((r>20)&(r<60))[0]
print (a)
print (b)
```

```python
#3. Given the following array,
#arr = np.array([14, 7, 25, 10, 32, 18, 5, 40, 22, 16])
#a. Extract all elements located at even indices.
#b. Reverse the extracted array
arr = np.array([14, 7, 25, 10, 32, 18, 5, 40, 22, 16])
print ("a",arr[1::2])
print ("b",np.sort(arr)[::-1])
```

```python
#5. Given the following 2D array,
#arr = np.array([[12, 5, 18],
# [7, 25, 10],
# [30, 14, 20]])
#a. Calculate the sum of each row and each column.
#b. Flatten the array into a one-dimensional array.
#c. Sort the flattened array in descending order.
s=np.array([[12,5,18],
             [7,25,10],
             [30,14,20]])
print (np.sum(s,axis=0))
print (np.sum(s,axis=1))
b=s.flatten()
print (np.sort(b)[::-1])
print(b)
```

```python
#4. Create the following two arrays,
#arr1 = np.array([10, 20, 30, 40])
#arr2 = np.array([50, 60, 70, 80])
#a. Combine them vertically into a 2D array.
#b. Split the resulting array into two separate arrays.
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([50, 60, 70, 80])
a=np.vstack((arr1,arr2))
b=np.vsplit(a,2)
print (a)
print (b)
```
