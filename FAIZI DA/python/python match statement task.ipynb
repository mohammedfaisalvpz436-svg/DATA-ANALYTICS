```python
#1. Display the full form of a department code using match ( CS → Computer
#Science, IT → Information Technology, EC → Electronics, ME → Mechanical)
department=str(input("enter department code"))
match department:
    case"cs":
        print ("computer science")
    case "it":
        print ("information technology")
    case "ec":
        print ("electronics")
    case "me":
        print ("mechanical")
        
```
