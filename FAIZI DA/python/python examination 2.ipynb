```python
#8. Display the day of the week based on a number from 1 to 7.
week=int(input("day number"))
match week:
  case 1:
      print ("monday")
  case 2:
      print ("tuesday")
  case 3:
      print ("wednesday")
  case 4:
      print ("thursday")
  case 5:
      print ("friday")
  case 6:
      print ("saturday")
  case 7:
      print ("sun day")
```
