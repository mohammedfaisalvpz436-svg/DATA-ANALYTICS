```python
#1. Print the numbers from 1 to 5 using a for loop.
for x in range (1,6):
    print (x)
```

```python
#2. Print the numbers from 10 to 6 using a while loop.
x=10
while x >=6:
    print (x)
    x-=1
```

```python
#3. Print the square of numbers from 1 to 5 using a for loop.
for x in range (1,6):
    print (x**2)
```

```python
#4. Print all even numbers from 2 to 10 using a while loop.
x=2
while x<=10:
    print (x)
    x+=2
```

```python
#6. Print each element of the following list using a while loop:
fruits = ["apple", "banana", "mango"]
i=0
while i < len(fruits):
    print(fruits[i])
    i=i+1
```

```python
#7. Find the sum of numbers from 1 to 5 using a for loop.
total=0
for i in range(1,6):
    total +=i
print (total)
```

```python
#8. Count the numbers in the following list using a while loop:
numbers = [10, 20, 30, 40]
count=0
i=0
while i < len(numbers):
    count+=1
    i+=1
print (count)
    
```
