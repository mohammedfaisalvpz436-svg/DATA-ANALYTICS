```python
#1. Check whether a number is positive, negative, or zero1
num=int(input("number"))
if num>0:
    print ("positive")
elif num<0:
    print ("negative")
elif num == 0:
    print ("zero")
else:
    print ("none of the above")
```

```python
#2. Check whether a number is even or odd.
x=int(input("number"))
if x % 2==0:
    print ("even")
else:
    print ("odd")
```

```python
#4. Find the largest among three numbers.
x=int(input("no1"))
y=int(input("no2"))
z=int(input("no3"))
if x>=y and x>=z:
    print ("no1 is greater")
elif y>=x and y>=z:
    print ("no2 is greater")
else:
    print ("no3 is greater")
```

```python
#5. Check whether a number is divisible by both 3 and 5.
x=int(input("number"))
if x%3 == 0 and x%5 == 0:
    print ("divisible by 3,5")
else:
    print ("not divisible by 3,5")
```

```python
#6. Check whether a year is a leap year.
x=int(input("year"))
if x%400 == 0:
    print("leap year")
elif x%4==0:
    print ("leap year")
elif x%100==0:
    print ("leap year")
else:
    print ("not leap year")
```

```python
#9. Assign a grade based on marks (A, B, C, D, Fail).
x=int(input("mark"))
if x>=90:
    print ("A")
elif x>=80:
    print ("B")
elif x>=70:
    print ("C")
else:
    print ("D")
```

```python
#14. Create a simple calculator using the user's choice of operator (+, -, *, /).

a= int(input("number"))
b=int(input("2nd number"))
operator =input("enter the operator (+,-,*,/) :")
if operator=="+":
    print ("result=",a+b)
elif operator=="-":
    print ("result=",a-b)
elif operator=="*":
    print ("result",a*b)
elif operator=="/":
    print ("result",a/b)
else:
    print ("invalid")
```
