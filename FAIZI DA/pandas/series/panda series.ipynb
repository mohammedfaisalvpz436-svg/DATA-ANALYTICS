```python
import pandas as pd
```

### series

```python
a=pd.Series([10,20,30,40,50])
print (a)
```

```python
# series from list of string
s=pd.Series(["apple","banana","mango","orange"])
print (s)
```

```python
# custom index
a=pd.Series(
    [100,200,300,400],
    index=["a","b","c","d"]
)
print (a)
```

```python
#series from a dictionary 
q={
    "a":100,
    "b":200,
    "c":300,
    "d":400,
}
a=pd.Series(q)
print (a)
```

```python
#iloc
a=pd.Series([10,20,30,40,50])
print (a.iloc[0])
```

```python
#loc
a=pd.Series(
    [100,200,300,400],
    index=[2,1,4,5]
)
print (a.loc[1])
```

```python
#slicing a series
a=pd.Series ([10,20,30,40,50])
print (a.loc[:3])
```

```python
#first and last elements 
#first
print (a.iloc[0])
#last
print (a.iloc[-1])
#first 3
print (a.iloc[:3])
#last two
print (s.iloc[-2:])
```

```python
#arithematic operations 
a=pd.Series([10,20,30,40,50])
print (a+5)
```

```python
#mathematical functions
mark=pd.Series([75,88,92,65,80])
print (mark.sum())
print (mark.mean())
print (mark.max())
print (mark.min())
```

```python
#statistical method
a=pd.Series([10,20,30,40,50])
print (a.sum())
print (a.mean())
print (a.median())
print (a.max())
print (a.min())
```

# Data Selection Using loc and iloc

```python
import pandas as pd
```

```python
data={
    "name":["asha","rahul","anu"],
    "age":[20,21,19],
    "city":["kochi","delhi","chennai"]
}
df=pd.DataFrame(data)
print (df)
```

```python
print (df.loc[1])
```

```python
print (df.loc[0:1,["name","age"]])
```

```python
print (df.iloc[1])
```

```python
print (df.iloc[0:2,0:2])
```

```python
#difference in slicing
print (df.loc[0:1])
```

```python
print (df.iloc[0:1])
```

```python
print (df.loc[2,"city"])
```

```python
print (df.iloc[2,2])
```
