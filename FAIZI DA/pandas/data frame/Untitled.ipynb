```python
import pandas as pd
```

## data frame

```python
data ={
    "name":["anu","rahul","priya"],
    "age":[21,22,20],
    "mark":[85,90,88]
}
df=pd.DataFrame(data)
print (df)
```

## creating a data frame a list

```python
lists=[
    ["anu",21,85],
    ["rahul",22,90],
    ["priya",20,88]
]
df=pd.DataFrame(lists,columns=["name","age","mark"])
print (df)
```

```python
## creating a dataframe with a custom index
data={
    "name":["anu","rahul","priya"],
    "mark":[85,90,88]
}
df=pd.DataFrame(data,index=["s1","s2","s3"])
print (df)
```

## display columns-df.columns

## adding a new column

```python
df=pd.DataFrame({
    "name":["anu","rahul","priya"],
    "mark":[85,90,88]
})
df["grade"]=["A","A+","A"]
print (df)
```

## modifying an existing column

```python
df["mark"]=[90,95,92]
print (df)
```
